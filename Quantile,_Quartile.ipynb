```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('Titanic Dataset.csv')
df.head()
```

```python
df.drop(columns=['Cabin', 'PassengerId', 'Ticket', 'Name'], inplace=True)
display(df.head())
```

```python
#Quantilesof age
age_q1=np.quantile(df['Age'],0.25)
print("First quantile: ", age_q1)
age_q2=np.quantile(df['Age'],0.50)
print("First quantile: " ,age_q2)
age_q3=np.quantile(df['Age'],0.75)
print("First quantile: ", age_q3)
```

```python
#interquartile of age
age_inqr = age_q3 - age_q1
print("Age interquartile: ", age_inqr)
```

```python
#Quantiles of Fare
fare_q1=np.quantile(df['Fare'], 0.25)
print("Quantile 1 of Fare: ", fare_q1)
fare_q2=np.quantile(df['Fare'], 0.50)
print("Quantile 2 of Fare: ", fare_q2)
fare_q3=np.quantile(df['Fare'], 0.75)
print("Quantile 3 of Fare: ", fare_q3)
```

```python
fare_iqr = fare_q3 - fare_q1
print("Fare Interquartile: ", fare_iqr)
```

```python
sns.boxplot(df['Fare'], palette='rainbow', legend=True)
plt.title("Fare Distribution")
```

```python
#boxplot
plt.boxplot(df['Fare'])
plt.title("Fare Distribution")
plt.show()
```

```python
sns.boxplot(df['Age'], palette='turbo', legend=True)
plt.title("Age Distribution")
```
